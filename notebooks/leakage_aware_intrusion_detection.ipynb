## Why rebuild the original analysis?

The original coursework explored CIC-IDS2017 thoroughly, but preprocessing and SMOTE occurred before data splitting. This version keeps every learned transformation inside the training pipeline, so the held-out test data remains unseen.


```python
from scripts.generate_demo import synthetic_flows
from cicids2017 import train_and_evaluate

frame = synthetic_flows()
frame.head()
```

The synthetic data below exists only to exercise the software. Its scores are not CIC-IDS2017 research results.


```python
frame['Label'].value_counts()
```

```python
binary = train_and_evaluate(frame, task='binary')
{'balanced_accuracy': binary.balanced_accuracy, 'macro_f1': binary.macro_f1}
```

```python
multiclass = train_and_evaluate(frame, task='multiclass')
{'balanced_accuracy': multiclass.balanced_accuracy, 'macro_f1': multiclass.macro_f1}
```

```python
import pandas as pd
pd.DataFrame(multiclass.confusion_matrix, index=multiclass.labels, columns=multiclass.labels)
```
